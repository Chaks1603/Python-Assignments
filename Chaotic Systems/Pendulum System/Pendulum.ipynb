The **fourth-order Runge-Kutta** algorithm, **$\textbf{RK4}$**, has proven to be robust and capable of industrial strength work.                                 
The algorithm is based upon the formal integral of the differential equation:                                                          
$\qquad\begin{aligned} \frac{dx}{dt} &= f(t, x) \\  x(t) &= \int f(t, x) \ dt \\  x_{n+1} &= x_n + \int^{t_{n+1}}_{t_n} f(t, x) \ dt  \end{aligned}$

We will always use the **4th order Runge-Kutta method**, which in vector form is

$\quad\displaystyle\begin{equation} \mathbf{x}_{n+1} = \mathbf{x}_n + \frac{\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4}{6} \end{equation}$

where                                                                                                                               
$\qquad\begin{aligned} \mathbf{k}_1 &=  \mathbf{f}(\mathbf{t}_n,\mathbf{x}_n) \ \mathrm{dt}  \\                              \mathbf{k}_2 &= \mathbf{f}\big(\mathbf{t}_n + \frac{\mathrm{dt}}{2},\mathbf{x}_n + \frac{\mathbf{k}_1}{2}\big) \ \mathrm{dt} \\ \mathbf{k}_3 &= \mathbf{f}\big(\mathbf{t}_n + \frac{\mathrm{dt}}{2},\mathbf{x}_n + \frac{\mathbf{k}_2}{2}\big) \  \mathrm{dt}  \\ \mathbf{k}_4 &= \mathbf{f}\big(\mathbf{t}_n + \mathrm{dt},\mathbf{x}_n + \mathbf{k}_3\big) \  \mathrm{dt}  \end{aligned}$

The critical idea here is to expand the integrand around the midpoint of the integration interval and thereby obtain  
$O(h^4)$ precision via the cancellation of the $h$ and $h^3$ terms .                                        
The price paid for the improved precision is having to approximate three derivatives and the unknown $x$ at the middle of the interval.

```python
def RK4(f, t_span, const, dt, x0):
    time   = np.arange(t_span[0], t_span[1], dt)
    x_n    = np.zeros((len(time), len(x0)), dtype=float)
    x_n[0] = x0
    
    for i in range(len(time)-1):
        k1 = dt * np.array(f(time[i]         , x_n[i]         , const))
        k2 = dt * np.array(f(time[i] + 0.5*dt, x_n[i] + 0.5*k1, const))
        k3 = dt * np.array(f(time[i] + 0.5*dt, x_n[i] + 0.5*k2, const))
        k4 = dt * np.array(f(time[i] +     dt, x_n[i] +     k3, const))
        x_n[i+1] = x_n[i] + (k1 + 2*k2 + 2*k3 + k4) / 6
    return time, x_n
```

### $\text{Damped Driven Pendulum's Oscillation Equation}$                                                                                                      
The equation of motion is just $I\ddot{\theta} = \Gamma$, where $I$ is the moment of inertia and $\Gamma$ is the net torque about the pivot.                                                                                                                     
In this case $I = m L^2 ,$ and the torque arises from the three forces shown in Figure .    
<img src="https://raw.githubusercontent.com/Chaks1603/Storage/main/Driven-damped-pendulum.png" alt="Driven-damped-pendulum" style="float: right; width: 250px;"/>The resistive force has magnitude $bv$ and hence exerts a torque $-Lbv = -bL^2\dot{\theta}$.                                    
The torque of the weight is $-mgL\sin\theta,$ and that of the driving force is $LF(t)$.                                             
Thus the equation of motion $I\ddot{\theta}=\Gamma$ is                                                                                   
$\qquad\begin{aligned} mL^2\ddot{\theta}=-bL^2\dot{\theta}-mgL\sin\theta+LF(t)   \end{aligned}$                                                                          
Let us assume that the driving force $F(t)$ is sinusoidal $\begin{aligned} ,{i.e.}, \quad F(t)=F_o\cos(\omega t)   \end{aligned}$                                                                    
where $F_o$ is the drive amplitude (the amplitude of the driving force) and $\omega$ the drive frequency.                         
Substituting $(2)$ in $(1)$, we find that,                                                                                                                   
$\quad\begin{aligned} &{\therefore}\quad mL^2\ddot{\theta}=-bL^2\dot{\theta}-mgL\sin\theta+LF_o\cos(\omega t)  \\ 
&{or}, \quad\ddot{\theta} + \underbrace{\frac{b}{m}}_{\equiv \ \displaystyle 2\beta} \dot{\theta} + \underbrace{\frac{g}{L}}_{\equiv \ \displaystyle\omega_o^2} \sin\theta = \underbrace{\frac{F_o}{mL}}_{\equiv \ \displaystyle\gamma\omega_o^2} \cos\omega t \qquad \bigg[\because \gamma=\frac{F_o}{mL\omega_o^2}=\frac{F_o}{mL \times \frac{g}{L}}=\frac{F_o}{mg} \bigg] \\
&{or}, \quad\ddot{\theta}+2\beta\dot{\theta}+\omega^2_o\sin{\theta} = \gamma\omega^2_o\cos\omega t  \\
&{\text{where },}\quad \text{ damping constant }=\beta,  \text{ natural frequency }=\omega_o, \text{ driving strength }=\gamma \ (\text{dimensionless}) \end{aligned}$




From here, you can proceed to find :                                                                                                                      
* Numerical integration using RK4 to solve this system for different parameters.                                                                                
* Phase space plots $ (\theta,\dot{\theta})$                                                                                                                
* Poincaré sections (sampled at every drive period)

* Bifurcation diagrams as you vary $\gamma$

* Lyapunov exponent calculations to analyze chaos.

```python
def damped_driven_pendulum(t, y, params):
    theta, theta_dot = y
    beta, omega_o, gamma, omega = params
    dtheta_dt   = theta_dot
    ddtheta_ddt = -2 * beta * theta_dot - omega_o**2 * np.sin(theta) + gamma * omega_o**2 * np.cos(omega * t)
    return [dtheta_dt, ddtheta_ddt]
```

$\begin{aligned} \qquad
e^{ix} &= \cos x + i\sin x \\

$ or, \quad \displaystyle \cos x = \frac{1}{2} \left( e^{ix} + e^{-ix} \right) $

Cubing above eqn. on both sides we get :                                                                                                    
$ \begin{aligned} or, \quad \cos^3 x &= \bigg(\frac{1}{2}( e^{ix} + e^{-ix}) \bigg)^3 \\ &= \frac{1}{8}\bigg( e^{ix} + e^{-ix} \bigg)^3 \\ &= \frac{1}{8}\bigg( e^{3ix} + e^{-3ix} + 3\cdot e^{ix} \cdot e^{-ix}\cdot( e^{ix} + e^{-ix})\bigg) \qquad \big[ \ \because \ (a+b)^3 = a^3+b^3+3\cdot a\cdot b\cdot(a+b)\big] \\ &= \frac{1}{8}\bigg( (e^{3ix} + e^{-3ix}) + 3( e^{ix} + e^{-ix})\bigg) \\ &= \frac{1}{4}\bigg( \frac{1}{2} \underbrace{(e^{3ix} + e^{-3ix})}_{\equiv \ 2\cos 3x} + \frac{3}{2} \underbrace{( e^{ix} + e^{-ix})}_{\equiv \ 2\cos x} \bigg)  \\ &= \frac{1}{4} (\cos{3x} + 3\cos{x}) \end{aligned}$

#  Trying to find the Bifurcation points of $\gamma$

## Step 1: Mapping Spatial Ensembles to Parameter Fields
Let $\Gamma = \{\gamma_1, \gamma_2, \dots, \gamma_M\}$ be a well-ordered set of control parameters ($\gamma_i \in \mathbb{R}$).                                
Let $\Theta_i = \{\theta^{(1)}, \theta^{(2)}, \dots, \theta^{(K)}\}$ be the set of all steady-state angular positions observed on the Poincaré section at parameter $\gamma_i$.                                                                                                                                   
We define an aggregation operator $\mathcal{A}: \Gamma \to \mathcal{P}(\mathbb{R})$ that maps each discrete parameter value to its corresponding set of state trajectories:                                                                                                                                        
$$\mathcal{A}(\gamma_i) = \bigcup_{k=1}^{K} \theta^{(k)}$$                                                                                                   
## Step 2: Metric Quantization & Attractor Cardinality
Because numerical ODE integration contains floating-point noise, the raw set $\mathcal{A}(\gamma_i)$ is dense with near-identical values.                    
To isolate true periodic orbits, we project the continuous state space onto a coarse-grained grid using a quantization function $Q_\epsilon(x) = \lfloor \frac{x}{\epsilon} \rceil \cdot \epsilon$, where $\epsilon = 10^{-5}$.                                                                                 
The effective cardinality (or period size) $N(\gamma_i)$ of the attractor is defined as the count of distinct points under this quantization mesh:            
$$N(\gamma_i) = \left\vert{} \left\{ Q_\epsilon(\theta) \ \big\vert{}\ \theta \in \mathcal{A}(\gamma_i) \right\} \right\vert{}$$                                   
## Step 3: Discrete Bifurcation Operator
A bifurcation event signifies a topological mutation in phase space.                                                                                          
We define a forward-difference operator $\Delta N_i$ tracking structural changes across parameter increments:                                                  
$$\Delta N_i = N(\gamma_i) - N(\gamma_{i-1})$$                                                                                                              
A Period-Doubling Bifurcation Boundary occurs when the change matches a geometric doubling constraint:                                                       
$$\Delta N_i = N(\gamma_{i-1}) \quad \implies \quad \frac{N(\gamma_i)}{N(\gamma_{i-1})} = 2$$                                                               
A Chaotic Crisis Boundary is identified when a sudden topological expansion occurs over a small step size:                                                 
$$\Delta N_i > \tau \quad (\text{where } \tau = 10)$$ 

```python
def aggregate_theta_points_by_gamma(R, X):
    aggregated_theta = defaultdict(list)
    for r, x in zip(R, X):
        if isinstance(x, (np.ndarray, list)):
            aggregated_theta[r].append(x)       
        else:
            aggregated_theta[r].append([x])  
    return aggregated_theta

def find_max_theta_points_per_gamma(aggregated_theta):
    theta_max_per_gamma = {}
    for r, theta_points in aggregated_theta.items():
        all_theta_values = []
        for points in theta_points:
            if isinstance(points, np.ndarray):
                all_theta_values.extend(points)  
            else:
                all_theta_values.extend(np.array(points))         
        all_theta_values       = np.array(all_theta_values)
        unique_theta_values    = np.unique(np.round(all_theta_values, decimals=5))
        num_unique             = len(unique_theta_values)        
        theta_max_per_gamma[r] = num_unique    
    return theta_max_per_gamma

def detect_bifurcation_points(max_theta_points_per_gamma):
    bifurcation_info = []
    sorted_gamma = sorted(max_theta_points_per_gamma.keys())    
    for i in range(1, len(sorted_gamma)):
        prev_gamma    = sorted_gamma[i - 1]
        current_gamma = sorted_gamma[i]        
        prev_point    = max_theta_points_per_gamma[prev_gamma]
        current_point = max_theta_points_per_gamma[current_gamma]
        if abs(current_point - prev_point) > 10: 
            bifurcation_info.append((current_gamma, current_point)) 
    return bifurcation_info

def find_local_maxima(bifurcation_info):
    local_maxima = []
    for i in range(1, len(bifurcation_info) - 1):
        prev_gamma, prev_points = bifurcation_info[i - 1]
        curr_gamma, curr_points = bifurcation_info[i]
        next_gamma, next_points = bifurcation_info[i + 1]
        if curr_points > prev_points and curr_points > next_points:
            local_maxima.append((curr_gamma, curr_points))
    return local_maxima
```

```python
def fit(x,y):
    n   = len(x)
    sx  = sum(x)
    sy  = sum(y)
    sx2 = sum(x*x)
    sxy = sum(x*y)
    b   = (n*sx2 - sx**2)
    m   = (n*sxy - sx*sy) / b
    c   = (sy*sx2 - sx*sxy) / b
    return m, c
```

## Deep Study of DPP Exponentials - Lyapunov's 

### Step 1: Phase Space Dynamics & Perturbation Vector
The Driven Damped Pendulum (DDP) is described by a non-autonomous vector field $\dot{\mathbf{x}} = \mathbf{F}(t, \mathbf{x}; \mathbf{p})$,                   
where $\mathbf{x}(t) = [\theta(t), \omega(t)]^T$ represents the state vector, and $\mathbf{p} = [\beta, \omega_0, \gamma, \Omega]^T$ represents the parameter vector.                                                                                                                                             
To determine the sensitivity to initial conditions, we instantiate a fiducial trajectory $\mathbf{X}_1(t)$ and a shadow trajectory $\mathbf{X}_2(t)$,        
separated by an initial Euclidean perturbation vector $\mathbf{\delta}(0)$:                                                                               
$$\mathbf{X}_2(0) = \mathbf{X}_1(0) + \mathbf{\delta}(0), \quad \Vert{}\mathbf{\delta}(0)\Vert{} = d_0$$ 
### Step 2: Temporal Evolution via Runge-Kutta 4 (RK4)
Both state vectors are integrated over a discrete time step $\Delta t$ using a fourth-order accurate mapping $\Phi_{\Delta t}(\mathbf{x}) \approx \mathbf{x}(t+\Delta t)$.                                                                                                                                     
For any step $n$:                                                                                                                                           
$\qquad\begin{aligned}
\mathbf{k}_1 &= \mathbf{F}(t_n, \mathbf{x}_n) \cdot \Delta t \\ 
\mathbf{k}_2 &= \mathbf{F}\left(t_n + \frac{\Delta t}{2}, \mathbf{x}_n + \frac{\mathbf{k}_1}{2}\right) \cdot \Delta t \\
\mathbf{k}_3 &= \mathbf{F}\left(t_n + \frac{\Delta t}{2}, \mathbf{x}_n + \frac{\mathbf{k}_2}{2}\right) \cdot \Delta t \\
\mathbf{k}_4 &= \mathbf{F}(t_n + \Delta t, \mathbf{x}_n + \mathbf{k}_3) \cdot \Delta t \\
\mathbf{x}_{n+1} &= \mathbf{x}_n + \frac{1}{6}(\mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4) \end{aligned}$                                                
### Step 3: Gram-Schmidt Spatial Alignment & Rescaling
After each time increment $\Delta t$, the unnormalized physical separation is calculated:
$$d_n = \Vert{}\mathbf{X}_2(t_n) - \mathbf{X}_1(t_n)\Vert{}$$ 
Because chaotic orbits diverge exponentially ($d_n \propto d_0 e^{\lambda t}$),                                                                                 
leaving the shadow trajectory unchecked would quickly lead to floating-point overflow and saturation of the attractor bounds.                                                                                                        
To prevent this, the shadow trajectory is continuously projected back along the direction of alignment to a distance of exactly $d_0$:
$$\mathbf{X}_2^{\text{renorm}}(t_n) = \mathbf{X}_1(t_n) + d_0 \frac{\mathbf{X}_2(t_n) - \mathbf{X}_1(t_n)}{d_n}$$ 
### Step 4: Ergodic Limit Estimation
The Maximal Lyapunov Exponent $\lambda_{\max}$ is defined analytically as the infinite-time limit of the average growth rate.                                
Computationally, it is approximated over $M$ finite iterations following a transient warm-up phase:
$$\lambda_{\max} \approx \frac{1}{M \cdot \Delta t} \sum_{n=1}^{M} \ln\left(\frac{d_n}{d_0}\right)$$ 

```python
def rk4_step(f, t, y, dt, params):
    y = np.asarray(y, dtype=float)
    k1 = dt * np.asarray(f(t, y, params), dtype=float)
    k2 = dt * np.asarray(f(t + dt/2, y + k1/2, params), dtype=float)
    k3 = dt * np.asarray(f(t + dt/2, y + k2/2, params), dtype=float)
    k4 = dt * np.asarray(f(t + dt, y + k3, params), dtype=float)
    return y + (k1 + 2*k2 + 2*k3 + k4)/6
    
def lyapunov_ddp(beta, omega0, gamma, omega,dt=0.01,transient=5000,iterations=50000,d0=1e-8):
    params = [beta, omega0, gamma, omega]
    X1 = np.array([-np.pi/2, 0.0])
    X2 = X1 + np.array([d0, 0])
    t = 0
    for _ in range(transient):
        X1 = rk4_step(damped_driven_pendulum, t, X1, dt, params)
        X2 = rk4_step(damped_driven_pendulum, t, X2, dt, params)
        t += dt
        diff = X2 - X1
        d = np.linalg.norm(diff)
        X2 = X1 + d0 * diff/d
    lyap_sum = 0.0
    for _ in range(iterations):
        X1 = rk4_step(damped_driven_pendulum, t, X1, dt, params)
        X2 = rk4_step(damped_driven_pendulum, t, X2, dt, params)
        t += dt
        diff = X2 - X1
        d = np.linalg.norm(diff)
        lyap_sum += np.log(d/d0)
        X2 = X1 + d0 * diff/d
    return lyap_sum/(iterations*dt)

omega = 2*np.pi
omega0 = 1.5*omega
beta = omega0/4
gamma_values = np.linspace(1.06, 1.087, 1000)
lyap_values = []
for gamma in tqdm(gamma_values):
    lam = lyapunov_ddp(beta, omega0, gamma, omega,dt=0.01,transient=5000,iterations=30000)
    lyap_values.append(lam)

plt.figure(figsize=(10,5))
plt.plot(gamma_values,lyap_values,'-',ms=0.5)
plt.xlabel(r'$\gamma$')
plt.ylabel(r'Largest Lyapunov Exponent')
plt.title('DDP Largest Lyapunov Exponent')
plt.xticks(gamma_v,[fr'$\gamma_{i+1}$'for i in range(len(gamma_v))], fontsize=12)
plt.grid(True)
plt.show()
```

### $\text{COMPUTATIONAL METHODS FOR PHYSICS - JOEL FRANKLIN}$

Problem 13.12                                                                                                                                                      
A damped, driven pendulum can be described by the equation of motion (see, for
example [39]):

$ \qquad \displaystyle\ddot{\theta}=-\frac{g}{L}\sin\theta-\alpha\dot{\theta}+\beta\cos\left(\omega t\right)$

where we start with (6.17) and add a damping term $\thicksim-\dot{\theta}$ and time-dependent forcing function by hand with constants $\alpha$ and $\beta$ setting the strength of these terms.                                                                                                                       
If we set
$t = \sqrt{L/g}s$ for dimensionless $s$, then

$ \qquad \displaystyle\frac{d^2\theta}{ds^2}=-\sin\theta-\underbrace{\alpha\sqrt{\frac{L}{g}}}_{\equiv\text{ A}}\dot{\theta}+\underbrace{\beta\frac{L}{g}}_{\equiv\text{ B}}\cos\left(\omega \sqrt{\frac{L}{g}}s \right)$

and define the dimensionless driving frequency  $\Omega \equiv \omega\sqrt{\frac{L}{g}}$                                                                                   
(a) For $A = \frac{1}{0.5}, B = 1.5$ and $\Omega = \frac{2}{3},$ take $\Delta t = \frac{\Omega}{64},$ and run your RK4 ODE solver for $2^{16}$ steps starting from $\theta(0) = \frac{\pi}{2}$ and $\dot{\theta} = 0$.                                                            
Taking just the second half of the numerical solution (to allow transients to decay), plot the angle $\theta$ as a function of dimensionless time $s$.                                                                                                 
Compute the power spectrum and plot the log of the power spectrum, focusing on the area around the driving frequency $\Omega$.                                                                                                                            
(b) Re-do part (a) with $A = \frac{1}{3}$, in the chaotic region for this problem

```python
from scipy.fftpack import fft, fftfreq

def damped_driven_pendulum(s, state, params):
    theta, omega = state  # state = [theta, omega]
    A, B, Omega = params
    dtheta_ds = omega
    domega_ds = -np.sin(theta) - A * omega + B * np.cos(Omega * s)
    return np.array([dtheta_ds, domega_ds])

A_values = [1/0.5, 1/3]  # Two cases: non-chaotic and chaotic
B = 1.5
Omega = 2/3
dt = Omega / 64
num_steps = 2**16
s_span = (0, num_steps * dt)  # Time range
s_half = num_steps // 2       # Index for second half
theta0 = np.pi / 2
omega0 = 0
x0 = np.array([theta0, omega0])

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, A in enumerate(A_values):
    s_values, sol = RK4(damped_driven_pendulum, s_span, (A, B, Omega), dt, x0)
    s_values = s_values[s_half:]
    theta_values = sol[s_half:, 0]

    axes[i].plot(s_values, theta_values, color='b')
    axes[i].set_xlabel(r"Dimensionless time $s$")
    axes[i].set_ylabel(r"Angle $\theta$")
    axes[i].set_title(f"$A={A:.2f}$, $B={B}$,"+r"$\Omega=\frac{2}{3}$")
    
    # Compute power spectrum
    theta_fft = fft(theta_values - np.mean(theta_values))
    power_spectrum = np.abs(theta_fft)**2
    freq = fftfreq(len(theta_values), d=dt)  # Frequency axis

    mask = freq > 0  # Ignore negative frequencies
    axes[i+2].plot(freq[mask], np.log10(power_spectrum[mask]), color='r')
    axes[i+2].axvline(Omega, color='k', linestyle='--', label=r"$\Omega$")
    axes[i+2].set_xlabel("Frequency")
    axes[i+2].set_ylabel("Log Power")
    axes[i+2].set_title(f"Power Spectrum ($A={A:.2f}$)")
    axes[i+2].legend()

plt.tight_layout()
plt.show()
```

## $Chaotic \ Dynamics  :  An \ Introduction - G.L.Baker \ and \ J.P.Gollub$

The **simple pendulum** is a classic nonlinear system described by the equation:

$\qquad \displaystyle \frac{d^2\theta}{dt^2} + \sin\theta = 0 $

This models a frictionless pendulum swinging under gravity. However, real-world pendulums are affected by:

* **Damping** due to air resistance or friction at the pivot,
* **External periodic driving forces** (e.g., from a motor or an oscillating mount).

To include these effects, we modify the equation:

$\qquad \displaystyle \frac{d^2\theta}{dt^2} + \left(\frac{1}{q}\right)\frac{d\theta}{dt} + \sin\theta = g\cos\left(\omega_{\tiny D} t\right)$

Introduce:

   * $ \displaystyle \omega = \frac{d\theta}{dt}$ (angular velocity),
   * $ \displaystyle \phi = \omega_{\tiny D} t$ (driving phase).

Then the system becomes:

$ \qquad\begin{aligned} \frac{d\theta}{dt} &=\omega \\  \frac{d\omega}{dt} &= g\cos\phi- \frac{\omega}{q} -\sin\theta  \\ \frac{d\phi}{dt} &=\omega_{\tiny D} \end{aligned} $

where

* $\theta(t)$               : angular displacement,
* $\displaystyle\frac{1}{q}$ : damping coefficient (inversely related to the quality factor),
* $g$                      : amplitude of the driving force,
* $\omega_{\tiny D}$               : driving frequency.

  
This **three-dimensional system** captures the full dynamics of the driven damped pendulum, including the periodic nature of the drive through $\phi$.

```python
def pendulum_system(t, state, const):
    theta, omega, phi = state
    q, g, omega_D     = const
    dtheta_dt = omega
    domega_dt = g * np.cos(phi) - (1/q) * omega - np.sin(theta) 
    dphi_dt   = omega_D
    return [dtheta_dt, domega_dt, dphi_dt]
```

### Bifurcation

```python
def strobe_indices(t, omega_D, skip_cycles=30, sample_cycles=30):
    T     = 2 * np.pi / omega_D
    start = skip_cycles * T
    end   = (skip_cycles + sample_cycles) * T
    times = np.arange(start, end, T)
    return [np.argmin(np.abs(t - s)) for s in times]
```

The **simple pendulum** is a classic nonlinear system described by the equation:

$\qquad \displaystyle \frac{d^2\theta}{dt^2} + \sin\theta = 0 $

This models a frictionless pendulum swinging under gravity. However, real-world pendulums are affected by:

* **Damping** due to air resistance or friction at the pivot,
* **External periodic driving forces** (e.g., from a motor or an oscillating mount).

To include these effects, we modify the equation:

$\qquad \displaystyle \frac{d^2\theta}{dt^2} + \left(\frac{1}{q}\right)\frac{d\theta}{dt} + \sin\theta = g\sin\left(\omega_{\tiny D} t\right)$

Introduce:

   * $ \displaystyle \omega = \frac{d\theta}{dt}$ (angular velocity),
   * $ \displaystyle \phi = \omega_{\tiny D} t$ (driving phase).

Then the system becomes:

$ \qquad\begin{aligned} \frac{d\theta}{dt} &=\omega \\  \frac{d\omega}{dt} &= g\sin\phi- \frac{\omega}{q} -\sin\theta  \\ \frac{d\phi}{dt} &=\omega_{\tiny D} \end{aligned} $

where

* $\theta(t)$               : angular displacement,
* $\displaystyle\frac{1}{q}$ : damping coefficient (inversely related to the quality factor),
* $g$                      : amplitude of the driving force,
* $\omega_{\tiny D}$               : driving frequency.

  
This **three-dimensional system** captures the full dynamics of the driven damped pendulum, including the periodic nature of the drive through $\phi$.

```python
def pendulum_system(t, state, const):
    theta, omega, phi = state
    q, g, omega_D     = const
    dtheta_dt = omega
    domega_dt = g * np.sin(phi) - (1/q) * omega - np.sin(theta) 
    dphi_dt   = omega_D
    return [dtheta_dt, domega_dt, dphi_dt]
```

## $\textbf{NON-LINEAR  DYNAMICS  AND  CHAOS} - \ Steven \ H. \ Strogatz$

**4.4.1** <span style="color:red">(Validity of overdamped limit)                                                                                                                           
Find the conditions under which it is valid to approximate the equation                                                                                        
$ \qquad\displaystyle mL^2\ddot{\theta}+b\dot{\theta}+mgL\sin{\theta} = \Gamma$                                                                                                                                        
by its overdamped limit $\quad\displaystyle b\dot{\theta}+mgL\sin{\theta} = \Gamma$     .

 Starting from the full equation (point mass at length $L$):

$ \qquad\displaystyle I\ddot\theta + b\dot\theta + K\sin\theta = \Gamma,$

where $I=mL^2$ and $K=mgL$.                                                                                                                                  
The overdamped approximation is

$ \qquad\displaystyle b\dot\theta + K\sin\theta = \Gamma, $

For motion with a characteristic angular frequency $\displaystyle\omega$ (so $\dot\theta \sim \omega\Delta\theta,\ \ddot\theta\sim\omega^2\Delta\theta$), require

$ \qquad\displaystyle I\omega^2\Delta\theta \ll b\omega\Delta\theta \quad\Longrightarrow\quad I\omega \ll b.$

If the relevant frequency is the small-oscillation (natural) frequency $\displaystyle\omega_0=\sqrt{K/I}=\sqrt{g/L}$,                                                       
the condition becomes $ \quad\displaystyle I\omega_0 \ll b \quad\Longrightarrow\quad b \gg I\omega_0 = mL^2\sqrt{\frac{g}{L}} = m\sqrt{gL^3}.$

So **overdamped limit**  holds when $ \quad\displaystyle \boxed{\,b \gg m\sqrt{gL^3}\,}$

(or more precisely $\displaystyle b\gg 2m\sqrt{gL^3}$ if we want the usual damping-ratio factor of 2).

This can also be written in terms of the damping ratio $\zeta$,

$ \qquad\displaystyle\zeta=\frac{b}{2\sqrt{IK}}=\frac{b}{2\sqrt{mL^2\cdot mgL}}=\frac{b}{2m\sqrt{gL^3}}\qquad\text{require }\zeta\gg1.$

Let the inertial timescale be

$ \qquad\displaystyle\tau_{\rm in}=\frac{I}{b}=\frac{mL^2}{b}. $

If the actual angular motion varies on a timescale $T$,                                                                                                   
then $\displaystyle I\ddot\theta$ is small compared with $b\dot\theta$ whenever $ \displaystyle T \gg \tau_{\rm in}$                                                                                                                             
(i.e. the motion is slow compared with the inertial time).

So if the external forcing or initial perturbation produces slow evolution (large $T$),                                                                      
inertia can be dropped even if $b$ is not extremely large.

A convenient dimensionless measure is

$ \qquad\displaystyle\varepsilon \;=\; \frac{IK}{b^2}=\frac{mL^2\cdot mgL}{b^2}= \frac{m^2 g L^3}{b^2}.$

The overdamped approximation is valid when $\varepsilon\ll1$ (equivalently $b^2\gg m^2gL^3$).

**4.4.4** <span style="color:red">(Torsional spring)                                                                                                                            
Suppose that our overdamped pendulum is connected to a torsional spring.                                                                                   
As the pendulum rotates, the spring winds up and generates an opposing torque $-k\theta$.                                                                              
Then the equation of motion becomes                                                                                                                              
$\quad\displaystyle b\dot{\theta}+mgL\sin{\theta} = \Gamma-k\theta$.                                                                                             
a) Does this equation give a well-defined vector field on the circle?                                                                                                                              
b) Nondimensionalize the equation.                                                                                                                              
c) What does the pendulum do in the long run?                                                                                                                              
d) Show that many bifurcations occur as $k$ is varied from $0$ to $\infty$. What kind of bifurcations are they $?$

The system is described by                                                                                                                                    
$\qquad\displaystyle b \dot{\theta} + mg L \sin \theta = \Gamma - k \theta$                                                                                
where,                                                                                                                                                          
* $b$ = damping constant,
* $mgL \sin \theta$ = gravitational torque,
* $\Gamma$ = constant external torque,
* $k \theta$ = torsional spring torque.

**(a)** We can rewrite the equation as

$\qquad\qquad\displaystyle \dot{\theta} = \frac{\Gamma - k \theta - mg L \sin \theta}{b} $

* The vector field is $F(\theta) = \dot{\theta}(\theta)$.
* For a **well-defined vector field on the circle**, $F(\theta)$ must be **periodic** with period $2\pi$, because $\theta$ lives on the circle.

Here, $-k \theta$ is **not $2\pi$-periodic**, because $k(\theta + 2\pi) \neq k \theta$.

**Hence**, the vector field **is not well-defined on the circle** for $k \neq 0$, because the torsional spring term breaks $2\pi$-periodicity.

**(b)** Let’s define dimensionless variables

$\qquad\qquad\displaystyle\tau = \frac{mg L}{b} t, \quad \tilde{\Gamma} = \frac{\Gamma}{mg L}, \quad \tilde{k} = \frac{k}{mg L}$

Then,                                                                                                                                                        
$\qquad\qquad\displaystyle b \frac{d\theta}{dt} + mg L \sin \theta = \Gamma - k \theta \quad \Rightarrow \quad \frac{d\theta}{d\tau} + \sin\theta = \tilde{\Gamma} - \tilde{k} \theta $

or equivalently,   $\quad\displaystyle\boxed{\frac{d\theta}{d\tau} = \tilde{\Gamma} - \tilde{k} \theta - \sin\theta}$                                            
* This is a **dimensionless form**.

**(c)** This is an **overdamped system**, so it cannot oscillate. Long-term behavior is determined by **fixed points**:

$\qquad\qquad\displaystyle \dot{\theta} = 0 \quad \Rightarrow \quad \sin\theta + \tilde{k} \theta = \tilde{\Gamma} $

* This is a **transcendental equation**; solutions depend on $\tilde{k}$ and $\tilde{\Gamma}$.
* The system will relax to **one of these stable fixed points**.
* As $k$ increases, the spring dominates and the pendulum approaches $\theta \approx \Gamma/k$.
* So **in the long run, the pendulum settles to a static equilibrium** determined by the balance of spring, gravity, and external torque.

**(d)** Fixed points satisfy

$\qquad\qquad\displaystyle f(\theta) = \tilde{\Gamma} - \tilde{k} \theta - \sin\theta = 0 $

* For **small $k$**, gravity dominates → multiple fixed points possible (e.g., stable/unstable points near $\theta = n\pi$).
* As $k$ increases, the linear spring term $-\tilde{k}\theta$ dominates → eventually only **one fixed point** survives.

**Observation :**

* As $k$ increases, **pairs of fixed points collide and disappear**.

* This is a **saddle-node (fold) bifurcation**.

* So **many saddle-node bifurcations occur** as $k$ is increased from 0 to ∞.


```python
def pendulum_phase(t, state, const):
    theta = state[0]
    Gamma, k = const
    dtheta = Gamma - k*theta - np.sin(theta)
    return [dtheta]

theta_grid = np.linspace(-10,15,1000)
params = [ (1.2,0),(0.8,0),(np.pi,1), (np.pi/2,0.2)]
labels = [r'$\Gamma=1.2,\ k=0$',r'$\Gamma=0.8,\ k=0$',r'$\Gamma=\pi,\ k=1$',r'$\Gamma=\frac{\pi}{2},\ k=0.2$']
dt = 0.01
t_span = (0,15)
theta0_vals = np.linspace(-5,16,20)

for i,(Gamma,k) in enumerate(params):
   
    fig,axs = plt.subplots(1,2,figsize=(12,5))
    const = (Gamma,k)
    dtheta = Gamma - k*theta_grid - np.sin(theta_grid)

    axs[0].plot(theta_grid,dtheta)
    axs[0].set_xlabel(r'$\theta$')
    axs[0].set_ylabel(r'$\dot{\theta}$')
    axs[0].set_title(rf'Phase curve')
    axs[0].grid(True)

    for theta0 in theta0_vals:
        t, sol = RK4(pendulum_phase, t_span, const, dt, [theta0])
        theta_t = sol[:,0]
        axs[1].plot(t,theta_t,alpha=0.6)
    axs[1].set_xlabel('t')
    axs[1].set_ylabel(r'$\theta(t)$')
    axs[1].set_title('Time evolution')
    axs[1].grid(True)
    fig.suptitle(labels[i])
    plt.tight_layout()
    plt.show()
```

```python
def find_all_equilibria(Gamma, k_values, theta_search_grid):
    k_coords = []
    theta_roots = []
    stability_mask = []  # True for Stable, False for Unstable
    
    for k in k_values:
        f_val = Gamma - k * theta_search_grid - np.sin(theta_search_grid)
        sign_crossings = np.where(np.diff(np.sign(f_val)))[0]
        
        for idx in sign_crossings:
            # Fixed the library path call here
            root = sc.optimize.fsolve(lambda t: Gamma - k * t - np.sin(t), theta_search_grid[idx])[0]
            
            # Derivative test: f'(theta) < 0 means Stable, > 0 means Unstable
            derivative = -k - np.cos(root)
            
            k_coords.append(k)
            theta_roots.append(root)
            stability_mask.append(derivative < 0)
            
    return np.array(k_coords), np.array(theta_roots), np.array(stability_mask)

gamma_variants = [0.01, 7.50, 15.0]  # Three distinct driving torque strengths
k_space = np.linspace(0.005, 2.0, 10000)
theta_search_range = np.linspace(-5, 45, 25000)

fig, axs = plt.subplots(1, 3, figsize=(18, 5.5), layout="constrained")
fig.suptitle('Evolution of Saddle-Node Cascades Under Varying Driving Torques ($\Gamma$)', fontsize=16, fontweight='bold')

for idx, Gamma in enumerate(gamma_variants):
    ax = axs[idx]
    ks, thetas, is_stable = find_all_equilibria(Gamma, k_space, theta_search_range)
    ax.scatter(ks[is_stable], thetas[is_stable], color='b', s=1.0, 
               label='Stable (Sinks)' if idx == 0 else "", alpha=0.7)
    ax.scatter(ks[~is_stable], thetas[~is_stable], color='r', s=1.0, 
               label='Unstable (Saddles)' if idx == 0 else "", alpha=0.7)
    ax.set_title(rf'Low Driving: $\Gamma = {Gamma}$' if idx==0 else 
                 rf'Medium Driving: $\Gamma = {Gamma}$' if idx==1 else 
                 rf'High Driving: $\Gamma = {Gamma}$', fontsize=13)
    ax.set_xlabel(r'Spring Stiffness ($k$)', fontsize=11)
    ax.grid(True, linestyle=':', alpha=0.5)
    ax.set_xlim(0, 2.0)
    if len(thetas) > 0:
        ax.set_ylim(np.min(thetas) - 2, np.max(thetas) + 5)
    if idx == 0:
        ax.set_ylabel(r'Equilibrium Angles ($\theta^*$)', fontsize=12)
        ax.legend(markerscale=12, loc='upper right', fontsize=11)
    ax.text(0.35, Gamma + 3, 'Saddle-Node\nFolds', color='black', fontsize=9,
            bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray', boxstyle='round,pad=0.3'))
plt.show()
```

**6.6.5** <span style="color:red">Consider equations of the form $\displaystyle \ddot{x} + f(\dot{x}) + g(x)= 0$, where $f$ is an even function, and both $f$ and $g$ are smooth.                                                
a) Show that the equation is invariant under the pure time-reversal symmetry $t\rightarrow -t$.                                                                         
b) Show that the equilibrium points cannot be stable nodes or spirals.



$f(\dot{x})$ is **even** ($i.e., \ f(-v) = f(v)$)                                                                                                                   
 $f$ and $g$ are smooth.

#### **a) Time-reversal invariance**

Time-reversal symmetry means that if we replace $t \to -t$, the system remains unchanged in form.

 Let $t \to -t$.                                                                                                                                               
 Then,

$\qquad\qquad\displaystyle \frac{d}{dt} = \frac{d}{d(-t)} \frac{d(-t)}{dt} = -\frac{d}{d(-t)}$

* Velocity transforms: $\displaystyle\dot{x} = \frac{dx}{dt} \to \frac{dx}{d(-t)} = -\dot{x}$
* Acceleration transforms: $\displaystyle\ddot{x} = \frac{d\dot{x}}{dt} \to \frac{d(-\dot{x})}{d(-t)} = \ddot{x}$

 Substitute into the equation

$\qquad\qquad\displaystyle\ddot{x} + f(\dot{x}) + g(x) \to \ddot{x} + f(-\dot{x}) + g(x)$

 Since $f$ is **even**, $f(-\dot{x}) = f(\dot{x})$.                                                                                                            
 Therefore the equation remains exactly the same , $i.e.$,

$\qquad\qquad\displaystyle\ddot{x} + f(\dot{x}) + g(x) = 0 $

This proves the **system is invariant under pure time reversal** $t \to -t$.


#### **b) Equilibrium points cannot be stable nodes or spirals**

Equilibrium points occur when $\dot{x} = 0$ and $\ddot{x} = 0$.

$\qquad\qquad\displaystyle\ddot{x} = 0 \implies f(0) + g(x^*) = 0 \implies g(x^*) = -f(0)$

Let $x^*$ be an equilibrium point.

Linearize the system around $x^*$ by writing

$\qquad\qquad\displaystyle x = x^* + \xi$

* Then $\dot{\xi} = \eta$, $\ddot{\xi} = \dot{\eta}$
* Linearized equation:

$\qquad\qquad\displaystyle \ddot{\xi} + f'(\dot{x}^*) \dot{\xi} + g'(x^*) \xi = 0 $

* At equilibrium, $\dot{x}^* = 0 \implies f'(0)$ is **real**.                                                                                                     
  But since $f$ is **even**, $f'(0) = 0$ (derivative of even function at zero is zero).

$\qquad\qquad\displaystyle \implies \ddot{\xi} + g'(x^*) \xi = 0 $

 Linear system in standard form

$\qquad\qquad\displaystyle \dot{\xi} = \eta, \quad \dot{\eta} = -g'(x^*) \xi $

* Jacobian :

$\qquad\qquad\displaystyle J = \begin{bmatrix} 0 & 1 \\ -g'(x^*) & 0 \end{bmatrix} $

* Eigenvalues :

$\qquad\qquad\displaystyle \det{(J-\lambda I)}=\begin{vmatrix} -\lambda & 1 \\ -g'(x^*) & -\lambda \end{vmatrix} =0$

$\qquad\qquad\displaystyle \lambda^2 + g'(x^*) = 0 \implies \lambda = \pm i \sqrt{g'(x^*)} $

Eigenvalues are **purely imaginary**, which means the linearization is a **center**, not a node or spiral.

* No damping term exists at equilibrium ($f'(0) = 0$) because $f$ is even.
* Therefore, **stable nodes or spirals are impossible**.

**6.7.1** <span style="color:red">(Damped pendulum)                                                                                                                                 
Find and classify the fixed points of $\displaystyle\ddot{\theta}+b\dot{\theta}+\sin{\theta} = 0$ for all $b> 0$, and plot the phase portraits for the qualitatively different cases.

System $\quad\qquad\displaystyle \ddot{\theta} + b\dot{\theta} + \sin\theta = 0, 
\qquad b > 0.$

Let $\quad\displaystyle x = \theta, \quad y = \dot{\theta}.$

Then $\qquad\displaystyle \dot{x} = y, \qquad \dot{y} = -b y - \sin x.$

So the system is

$\qquad\qquad\displaystyle\begin{cases} \dot{x} = y, \\ \dot{y} = -b y - \sin x.\end{cases}$


### Fixed points

At a fixed point, $\dot{x} = 0$ and $\dot{y} = 0$.

So

$\qquad\qquad\displaystyle y = 0, \quad -\sin x = 0 \;\;\Longrightarrow\;\; x = n\pi, \; n \in \mathbb{Z}.$

Thus fixed points

$\qquad\qquad\displaystyle (x,y) = (n\pi, 0)$

### Jacobian

Jacobian at $(x,y)$

$\qquad\qquad\displaystyle J =
\begin{bmatrix}\frac{\partial \dot{x}}{\partial x} & \frac{\partial \dot{x}}{\partial y} \\ \frac{\partial \dot{y}}{\partial x} & \frac{\partial \dot{y}}{\partial y} \end{bmatrix}=\begin{bmatrix} 0 & 1 \\ -\cos x & -b \end{bmatrix}$

### Classification

Characteristic polynomial

$\qquad\qquad\displaystyle \lambda^2 + b\lambda + \cos x = 0. $

* **Case 1 : $x = 2n\pi$, so $\cos x = 1$.**

   $\qquad\qquad\displaystyle \lambda^2 + b\lambda + 1 = 0.$

   The roots 

   $\qquad\qquad\displaystyle\lambda = \frac{-b \pm \sqrt{b^2 - 4}}{2}.$

  * If $b^2 > 4$: two distinct real negatives → **stable node**.
  * If $b^2 = 4$: double root negative → **degenerate/critical node**.
  * If $b^2 < 4$: complex conjugates with negative real part → **stable spiral (focus)**.

   So $(2n\pi, 0)$ = **stable equilibrium**.

* **Case 2 : $x = (2n+1)\pi$, so $\cos x = -1$.**

   $\qquad\qquad\displaystyle \lambda^2 + b\lambda - 1 = 0.$

   The roots

   $\qquad\qquad\displaystyle \lambda = \frac{-b \pm \sqrt{b^2 + 4}}{2}$

   One root positive, one negative → **saddle point**.

   So $((2n+1)\pi, 0)$ = **unstable equilibrium**.


### Phase portraits 

* For $b > 0$, the stable equilibria at $(2n\pi,0)$ are **attractors** (node or spiral depending on $b$).
* The unstable ones at $((2n+1)\pi,0)$ are **saddles** with separatrices dividing basins of attraction.
* As damping increases, spirals turn into nodes.

**6.7.2** <span style="color:red">(Pendulum driven by constant torque)                                                                                                                  
The equation $\displaystyle \ddot{\theta}+\sin{\theta} = \gamma$                                                                                            
describes the dynamics of an undamped pendulum driven by a constant torque,                                                                                    
or an undamped Josephson junction driven by a constant bias current.                                                                                            
a) Find all the equilibrium points and classify them as $\gamma$ varies.                                                                                       
b) Sketch the nullclines and the vector field.                                                                                                                
c) Is the system conservative? If so, find a conserved quantity. Is the system reversible ?                                                                    
d) Sketch the phase portrait on the plane as $\gamma$ varies.                                                                                                     
e) Find the approximate frequency of small oscillations about any centers in the phase portrait.

Undamped pendulum driven by constant torque

$\qquad\qquad\displaystyle\ddot{\theta} + \sin\theta = \gamma$

Introduce

$\qquad\displaystyle \theta_1 = \theta, \quad \theta_2 = \dot{\theta} \implies
\dot{\theta}_1 = \theta_2, \quad \dot{\theta}_2 = \gamma - \sin\theta_1$

### (a) Equilibria and classification

**Equilibrium conditions :**

$\qquad\qquad\displaystyle \dot{\theta}_1 = 0 \implies \theta_2 = 0, \quad\dot{\theta}_2 = 0 \implies \sin\theta = \gamma$

* Fixed points exist only if $|\gamma| \le 1$.
* Equilibrium points:

$\qquad\qquad\displaystyle\theta^* = \arcsin\gamma + 2\pi n, \quad
\theta^* = \pi - \arcsin\gamma + 2\pi n, \quad n \in \mathbb{Z}
$

**Linearization :**

$\qquad\qquad\displaystyle J = \begin{pmatrix} 0 & 1 \\ -\cos\theta^* & 0 \end{pmatrix}, $

$\qquad\qquad\displaystyle \lambda^2 + \cos\theta^* = 0 \implies \lambda = \pm i \sqrt{\cos\theta^*}$

* $\theta^* = \arcsin\gamma \implies \cos\theta^* > 0$ → **center**
* $\theta^* = \pi - \arcsin\gamma \implies \cos\theta^* < 0$ → **saddle**
* $|\gamma| > 1$ → no fixed points → **rotational motion**

### (b) Nullclines and vector field

* $\dot{\theta}_1 = 0 \implies \theta_2 = 0$ → horizontal line
* $\dot{\theta}_2 = 0 \implies \sin\theta = \gamma$ → vertical line(s) if $|\gamma| < 1$

Vector field

$\qquad\qquad\displaystyle (\dot{\theta}, \dot{\theta}_2) = (\theta_2, \gamma - \sin\theta)$

### (c) Conserved quantity

Conservative system: Hamiltonian

$\qquad\displaystyle H(\theta, \dot{\theta}) = \frac{1}{2}\dot{\theta}^2 - \gamma \theta - \cos\theta = \text{constant}$

Reversible under $(t \to -t, \dot{\theta} \to -\dot{\theta})$

### (d) Phase portrait

* $|\gamma| < 1$: centers + saddles → closed orbits and separatrices
* $|\gamma| = 1$: saddle-node bifurcation → degenerate fixed point
* $|\gamma| > 1$: no fixed points → only rotational (open) trajectories

### (e) Frequency of small oscillations

Linearize near the center $\theta_c = \arcsin \gamma$:

$\qquad\qquad\displaystyle\ddot{\theta} \approx -\cos\theta_c (\theta - \theta_c)$

Angular frequency

$\qquad\qquad\displaystyle\omega = \sqrt{\cos\theta_c} = \sqrt{1 - \gamma^2}$

Small-oscillation frequency

$\qquad\qquad\displaystyle f = \frac{\omega}{2\pi} = \frac{\sqrt{1 - \gamma^2}}{2\pi}$

* Near the displaced equilibrium ($\theta_c$), the pendulum behaves like a **linear harmonic oscillator**.
* The restoring “stiffness” comes from gravity, reduced by the torque tilt
* As torque increases, oscillations **slow down** ($f$ decreases).
* At $|\gamma|=1$, oscillations vanish ($f=0$) because the torque exactly cancels the restoring force: the pendulum is neutrally balanced.

**6.7.3** <span style="color:red">(Nonlinear damping)                                                                                                                                    
Analyze $\displaystyle\ddot{\theta}+(1+a\cos\theta)\dot{\theta}+\sin\theta = 0$, for all $a\ge 0$.

We are given

$\qquad\qquad\displaystyle \ddot{\theta} + (1 + a \cos\theta)\dot{\theta} + \sin\theta = 0, \quad a \ge 0$

Introduce

$\qquad\qquad\displaystyle \theta_1 = \theta, \quad \theta_2 = \dot{\theta} $

Then

$\qquad\qquad\displaystyle \dot{\theta}_1 = \theta_2, \quad \dot{\theta}_2 = - (1 + a \cos\theta_1)\theta_2 - \sin \theta_1 $

So the system is

$\qquad\qquad\displaystyle \begin{cases} \dot{\theta} = \theta_2 \\ \dot{\theta}_2 = -\sin\theta - (1 + a \cos\theta)\theta_2 \end{cases} $

* Note: **damping depends on θ** → nonlinear damping.


Fixed points satisfy

$\qquad\qquad\displaystyle \dot{\theta} = 0 \implies \theta_2 = 0 $

$\qquad\qquad\displaystyle \dot{\theta}_2 = 0 \implies -\sin\theta - (1 + a \cos\theta) \cdot 0 = 0 \implies \sin\theta = 0$

So fixed points are independent of **a**:

$\qquad\qquad\displaystyle (\theta^*, \theta_2^*) = (n\pi, 0), \quad n \in \mathbb{Z}$

* Same as the linear damping case.


#### Jacobian

$\qquad\qquad\displaystyle J = \begin{pmatrix} 0 & 1 \\  -\cos\theta & -(1 + a \cos\theta)  \end{pmatrix} $

At $\theta^* = n\pi$

* Even $n$ ($\theta^* = 0, 2\pi, ...$) → $\cos\theta^* = 1$

$\qquad\qquad\displaystyle J = \begin{pmatrix} 0 & 1 \\ -1 & -(1 + a) \end{pmatrix} $

* Odd $n$ ($\theta^* = \pi, 3\pi, ...$) → $\cos\theta^* = -1$:

$\qquad\qquad\displaystyle J = \begin{pmatrix} 0 & 1 \\ 1 & -(1 - a) \end{pmatrix}$


#### Eigenvalues and classification

1. **Even n (θ = 0, 2π, ...)**

   Characteristic equation:

   $\qquad\qquad\displaystyle \lambda^2 + (1 + a)\lambda + 1 = 0 $

   * **Discriminant:** $\Delta = (1 + a)^2 - 4 = a^2 + 2a - 3$

    * $a < 1$: $\Delta < 0$ → **complex eigenvalues**, stable spiral
    * $a = 1$: $\Delta = 0$ → **degenerate node**
    * $a > 1$: $\Delta > 0$ → **stable node**

   * Always **stable** since real part = $-(1+a)/2 < 0$


2. **Odd n (θ = π, 3π, ...)**

   Characteristic equation

   $\qquad\qquad\displaystyle \lambda^2 + (1 - a)\lambda - 1 = 0$

   * **Discriminant:** $\Delta = (1 - a)^2 + 4 = a^2 - 2a + 5 > 0$ always
   * Eigenvalues: $ \displaystyle \lambda = \frac{-(1-a) \pm \sqrt{(1-a)^2 + 4}}{2}$

     * One positive, one negative → **saddle**
     * **Unstable** for all a ≥ 0


#### Effect of nonlinear damping

* The **stable node/spiral at θ = 0** becomes more overdamped as `a` increases.

* The **saddle at θ = π** remains a saddle, but the damping along the θ̇ direction is altered:

  * `a < 1`: weak damping along θ̇ direction
  * `a = 1`: one eigenvalue = 0 → marginal along one direction
  * `a > 1`: stronger damping along θ̇ direction

* Nonlinear damping **does not change the location of fixed points**, but changes the **rate at which trajectories approach/leave them**.


#### Phase portrait features

1. **Even n (stable) → trajectories spiral or move directly to θ = 0 depending on a**
2. **Odd n (saddle) → separatrices**
3. **a = 0 → linear damping**, standard spiral/node
4. **a > 0 → nonlinear damping → approach to stable points may be faster/slower depending on θ**

**6.7.4** <span style="color:red">(Period of the pendulum)                                                                                                                           
Suppose a pendulum governed by $\displaystyle\ddot{\theta}+\sin\theta = 0$ is swinging with an amplitude $\alpha$.                                        
Using some tricky manipulations, we are going to derive a formula for $T(\alpha)$, the period of the pendulum.                                                   
a) Using conservation of energy, show that $\displaystyle\dot{\theta}^2 = 2\left(\cos{\theta}−\cos{\alpha}\right)$ and hence that                                 
$\qquad\qquad\qquad\displaystyle T = 4\int_0^{\alpha} \frac{d\theta}{\left[ 2\left(\cos{\theta} - \cos{\alpha} \right) \right]^{1/2} }$                                       
b) Using the half-angle formula, show that                                                                                                                              
$\qquad\qquad\qquad\displaystyle T = 4\int_0^{\alpha} \frac{d\theta}{\left[ 4\left(\sin^2\frac{1}{2}{\alpha} - \sin^2\frac{1}{2}{\theta}\right) \right]^{1/2}}$                                                                                                                                                 
c) The formulas in parts (a) and (b) have the disadvantage that $\alpha$ appears in both the integrand and the upper limit of integration.                    
To remove the $\alpha -$dependence from the limits of integration, we introduce a new angle $\phi$ that runs from $0$ to $\frac{\pi}{2}$
when $\theta$ runs from $0$ to $\alpha$.                   
Specifically, let $\displaystyle \left(\sin\frac{1}{2}{\alpha}\right)\sin\phi= \sin\frac{1}{2}{\theta}$.                                                      
Using this substitution, rewrite (b) as an integral with respect to $\phi$. Thereby derive the exact result                                                     
$\qquad\qquad\qquad\displaystyle T = 4\int_0^{\pi/2} \frac{d\phi}{\cos\frac{1}{2}\theta} = 4K\left(\sin\frac{1}{2}{\alpha}\right)$,                                             
where the complete elliptic integral of the first kind is defined as                                                                                            
$\qquad\qquad\qquad \displaystyle K(m)=\int_0^{\pi/2} \frac{d\phi}{(1-m\sin^2{\phi})^{1/2}} \quad $ for $0\le m\le 1$.                                        
d) By expanding the elliptic integral using the binomial series and integrating term-by-term, show that                                                          
$\qquad\qquad\qquad\displaystyle T(\alpha) = 2\pi \left[1 + \frac{1}{16}\alpha^2 + \mathcal{O}(\alpha^4) \right] \quad $ for $\alpha\ll 1$.                       
Note that larger swings take longer.

### (a) Energy and the quadrature formula

Multiply $\quad\displaystyle\ddot\theta+\sin\theta=0$ by $\dot\theta$ and integrate:

$\qquad\displaystyle\dot\theta\ddot\theta+\dot\theta\sin\theta=0 \quad\Rightarrow\quad\frac{d}{dt}\!\Big(\tfrac12\dot\theta^2\Big)+\frac{d}{dt}(\!-\cos\theta)=0.$

Hence energy is conserved

$\qquad\displaystyle \tfrac12\dot\theta^2 -\cos\theta = E.$

For amplitude $\alpha$ we have $\dot\theta(\alpha)=0$ so $E=-\cos\alpha$. Thus,

$\qquad\displaystyle \dot\theta^2 = 2(\cos\theta-\cos\alpha)$

$\qquad\displaystyle \dot\theta = \sqrt{2(\cos\theta-\cos\alpha)}$

A quarter period is the time to go from $\theta=0$ to $\theta=\alpha$; hence the full period is

$\qquad\displaystyle\boxed{\,T(\alpha)=4\int_{0}^{\alpha}\frac{d\theta}{\sqrt{2(\cos\theta-\cos\alpha)}}\,.}$

### (b) Half-angle form

Using $\cos\theta=1-2\sin^2(\tfrac\theta2)$ to get

$\qquad\displaystyle \cos\theta-\cos\alpha =(1-2\sin^2\frac{\theta}{2})-(1-2\sin^2\frac{\alpha}{2}) =2\sin^2\frac{\alpha}{2}-2\sin^2\frac{\theta}{2} =2\Big(\sin^2 \frac{\alpha}{2}-\sin^2 \frac{\theta}{2}\Big) $

Substituting into the integrand yields

$\qquad\displaystyle\boxed{ \ T(\alpha)=4\int_{0}^{\alpha} \frac{d\theta}{\sqrt{2(\cos\theta-\cos\alpha)}}=4\int_{0}^{\alpha}\frac{d\theta}{\sqrt{2\cdot 2\Big(\sin^2 \frac{\alpha}{2}-\sin^2 \frac{\theta}{2}\Big)}} =4\int_{0}^{\alpha}\frac{d\theta}{\sqrt{4\Big(\sin^2 \frac{\alpha}{2}-\sin^2 \frac{\theta}{2}\Big)}} \ }$

### (c) Change of variable to remove $\alpha$ from the limits

Set $\quad\displaystyle\sin\frac{\theta}{2}=\sin\frac{\alpha}{2}\,\sin\phi,\qquad \phi\in[0,\tfrac{\pi}{2}].$

Differentiate w.r.t. $\phi$   


$\quad\displaystyle \frac{1}{2}\cos\frac{\theta}{2}\,\frac{d\theta}{d\phi} = \sin\frac{\alpha}{2}\,\cos\phi$


$\quad\displaystyle \frac{d\theta}{d\phi} = \frac{2\sin(\tfrac{\alpha}{2})\cos\phi}{\cos(\tfrac{\theta}{2})}$


Then $\quad\displaystyle d\theta = 2\frac{\sin(\alpha/2)\cos\phi}{\cos(\theta/2)}\,d\phi,$


We start from the substitution , $i.e.$, 

$\quad\displaystyle\sin\frac{\theta}{2} = \sin\frac{\alpha}{2}\,\sin\phi$

Square both sides

$\quad\displaystyle \sin^2\frac{\theta}{2} = \sin^2\frac{\alpha}{2}\,\sin^2\phi$

Subtract from $\sin^2(\alpha/2)$

$\quad\displaystyle\sin^2\frac{\alpha}{2} - \sin^2\frac{\theta}{2} = \sin^2\frac{\alpha}{2} - \sin^2\frac{\alpha}{2}\,\sin^2\phi = \sin^2\frac{\alpha}{2}\,(1 - \sin^2\phi) = \sin^2\frac{\alpha}{2}\,\cos^2\phi $

Since $\sin(\alpha/2) \ge 0$ and $\cos\phi \ge 0$ for $\phi\in[0,\pi/2]$,

$\quad\displaystyle \sqrt{\sin^2\frac{\alpha}{2}-\sin^2\frac{\theta}{2}} = \sin\frac{\alpha}{2}\,\cos\phi$

Substitute into the integrand of (b),

$\qquad\displaystyle \frac{d\theta}{\sqrt{4(\sin^2(\alpha/2)-\sin^2(\theta/2))}}
=\frac{2\frac{\sin(\alpha/2)\cos\phi}{\cos(\theta/2)}\,d\phi}{2 \,\sin(\alpha/2)\cos\phi}
=\frac{ \,d\phi}{\cos(\theta/2)}.$

Since, $\cos^2{\theta} + \sin^2{\theta} = 1 \implies \cos^2{(\tfrac{\theta}{2})} + \sin^2{(\tfrac{\theta}{2})} = 1 \implies \cos^2{(\tfrac{\theta}{2})} = 1 - \sin^2{(\tfrac{\theta}{2})} \implies \cos(\tfrac{\theta}{2})=\sqrt{1-\sin^2(\tfrac{\theta}{2})}$.                                  
Therefore

$\qquad\displaystyle T(\alpha)=4\int_0^{\pi/2}\frac{d\phi}{\cos(\tfrac{\theta}{2})}=4\int_0^{\pi/2}\frac{d\phi}{\sqrt{1-\sin^2(\tfrac{\theta}{2})}} =4\int_0^{\pi/2}\frac{d\phi}{\sqrt{1-\sin^2(\tfrac{\alpha}{2})\sin^2{\phi}}}$

Let $m=\sin^2\!\tfrac{\alpha}{2}$

Then 

$\qquad\displaystyle\boxed{ \ T(\alpha)=4\int_0^{\pi/2} \frac{d\phi}{\sqrt{1-m\sin^2{\phi}}}=4\,K\!\big(m\big)=4\,K\!\big( \sin^2\!\tfrac{\alpha}{2} \big)}$

so we obtain the standard elliptic form

where

$\qquad\displaystyle K(m)=\int_0^{\pi/2}\frac{d\phi}{\sqrt{1-m\sin^2\phi}}\qquad(0\le m\le1).$

### (d) Small-amplitude expansion  

For small $\alpha$ we have $\displaystyle m=\sin^2(\tfrac{\alpha}{2})\approx (\tfrac{\alpha}{2})^2=\tfrac{\alpha^2}{4}\ll1$.                             
The binomial expansion integrated termwise gives the standard series for $K(m)$

$\qquad\displaystyle K(m)=\frac{\pi}{2}\Big(1+\frac{m}{4}+\frac{9m^2}{64}+\cdots\Big).$

Thus

$\qquad\displaystyle T(\alpha)=4K(m)=2\pi\Big(1+\frac{m}{4}+\mathcal{O}(m^2)\Big)=2\pi\Big(1+\frac{1}{4}\cdot\frac{\alpha^2}{4}+\mathcal{O}(\alpha^4)\Big)$

so

$\qquad\displaystyle\boxed{\,T(\alpha)=2\pi\Big[1+\frac{1}{16}\alpha^2+\mathcal{O}(\alpha^4)\Big],\qquad \alpha\ll1.}$

This shows the period increases with amplitude; the first correction beyond $2\pi$ is $+\tfrac{1}{16}\alpha^2$.

For small $x$

so the equation becomes

$\qquad\displaystyle \ddot{x} + x - \frac{1}{6} x^3 = 0 \quad \implies \quad \ddot{x} + x = \frac{1}{6} x^3 $

We define the **perturbation** term as $\quad\displaystyle h(x) = -\frac{1}{6} x^3 $

so that the system is in the standard form

$\qquad\displaystyle \ddot{x} + x + \epsilon h(x) = 0, \quad \epsilon = 1$

Write $x(t)$ in amplitude-phase form:

$\qquad\displaystyle x(t) = r(T) \cos \theta(t), \quad \dot{x}(t) = -r(T) \sin \theta(t)$

where $r(T)$ and $\theta(t)$ vary slowly ($T = \epsilon t$).

#### Averaging equations

From the perturbation method:

$\qquad\displaystyle \frac{dr}{dT} = \frac{1}{2\pi} \int_0^{2\pi} h(r \cos \theta) \sin \theta \, d\theta$

$\qquad\displaystyle \frac{d\phi}{dT} = \frac{1}{2\pi r} \int_0^{2\pi} h(r \cos \theta) \cos \theta \, d\theta$

#### Compute amplitude change

$\qquad\displaystyle \frac{dr}{dT} = \frac{1}{2\pi} \int_0^{2\pi} \Big(-\frac{1}{6} r^3 \cos^3 \theta\Big) \sin \theta \, d\theta

$\implies r(T) = r_0 = a$ is **constant**, as expected for a conservative system.

#### Compute phase change

$\qquad\displaystyle r \frac{d\phi}{dT} = \frac{1}{2\pi} \int_0^{2\pi} \Big(-\frac{1}{6} r^3 \cos^3 \theta\Big) \cos \theta \, d\theta
= -\frac{r^3}{12\pi} \int_0^{2\pi} \cos^4 \theta \, d\theta $

$\qquad\displaystyle \int_0^{2\pi} \cos^4 \theta \, d\theta = \frac{3\pi}{4} \quad \implies \quad r \frac{d\phi}{dT} = -\frac{r^3}{16} \quad \implies \frac{d\phi}{dT} = -\frac{r^2}{16} = -\frac{a^2}{16}$

#### Compute frequency

The corrected frequency 

$\qquad\displaystyle \omega = 1 + \epsilon \frac{d\phi}{dT} \approx 1 - \frac{1}{16} a^2$

#### Physical interpretation

* The cubic term $-\frac{1}{6} x^3$ **reduces the restoring force** for larger amplitudes → oscillations **slow down**.
* This is consistent with the **softening spring** behavior.
* The frequency decreases quadratically with amplitude for small oscillations.

$\qquad\displaystyle  T = \frac{2\pi}{\omega}= \frac{2\pi}{1 - \frac{1}{16} a^2  + \mathcal{O}(a^4)}=2\pi\left[ 1 + \frac{1}{16}a^2 + \left( \frac{1}{16}a^2 \right)^2 + \left( \frac{1}{16}a^2 \right)^4 + \cdots \right] =  2\pi \left(1 + \frac{1}{16} a^2 + O(a^4)\right)$

**8.5.5** <span style="color:red">(Driven pendulum with quadratic damping)                                                                                                            
Consider a pendulum driven by a constant torque and damped by air resistance. In dimensionless form, the governing equation is                               
$\quad\displaystyle b\dot{\theta}+mgL\sin{\theta} = \Gamma-k\theta$.                                                                                             
where $\alpha \ge 0$ and $F \ge 0$ are the dimensionless damping strength and torque, respectively.                                                         
The new feature here is that the damping is **quadratic**, rather than linear, in the velocity $v = \dot{\theta}$.                                         
This is more realistic if the damping is primarily due to drag, but the trouble is that the damping becomes nonlinear, which normally would make the analysis harder.                                                                                                                                                 
But as you’ll see, the pleasant surprise here is that quadratic damping actually makes the system easier—in fact, it becomes explicitly solvable!                
**(a)** Find and classify the fixed points in the $(\theta, v)$ phase plane.                                                                               
If you find a center according to the linearization, decide whether this borderline case is truly a nonlinear center, a stable spiral, or an unstable spiral.                                                                                                                                                      
*(Hint : Find a local Lyapunov function in the neighborhood of the fixed point.)*                                                                            
**(b)** For $F > 1$, prove that the system has a stable limit cycle (where we now regard the phase space as a cylinder rather than a plane).               
Then prove that this limit cycle is **unique**.                                                                                                                
 Remarkably, exact formulas for the limit cycle and the homoclinic bifurcation curve can be found (**Pedersen and Saermark 1973**);                           
 this is one of the advantages of the quadratically damped case. However, the solution involves some tricky changes of variables.                               
**(c)** In the region $v > 0$, $\theta(t)$ increases monotonically. Therefore it can be inverted formally to yield $t(\theta)$.                         
Now regard $\theta$ as a new independent (time-like) variable, and introduce the new dependent variable
$\quad\displaystyle u = \frac{1}{2} v^2$                                                                                                                      
Use the chain rule (carefully, showing all your steps) to deduce that
$\quad\displaystyle \frac{du}{d\theta} = \ddot{\theta}$                                                                                                       
**(d)** Hence, in the region $v > 0$, the pendulum equation becomes                                                                                      
$\qquad\displaystyle \frac{du}{d\theta} + 2 \alpha u = F - \sin\theta$                                                                                       
which is a **linear equation in $u(\theta)$**.                                                                                                          
Assuming that this equation is correct (even if you were unable to derive it), find an **exact formula for the limit cycle** when $F > 1$.                     
**(e)** Now decrease $F$ while keeping $\alpha$ fixed. Show that the limit cycle undergoes a **homoclinic bifurcation** at some critical value of $F$,         
call it $F = F_c(\alpha)$, and give an **exact formula** for the bifurcation curve $F_c(\alpha)$.



## a) Fixed points and linear stability

**Step 1: Convert to first-order system**

Let $v = \dot{\theta}$, then

$\qquad\displaystyle\begin{cases} \dot{\theta} = v \\ \dot{v} = F - \alpha v |v| - \sin \theta \end{cases}$


**Fixed points**

$\qquad\displaystyle\dot{\theta} = 0 \Rightarrow v = 0$

$\qquad\displaystyle\dot{v} = 0 \Rightarrow F - \sin \theta = 0 \Rightarrow \theta^* = \arcsin(F), \quad \pi - \arcsin(F)$

Valid only if $F < 1$.

* Fixed points

$\qquad\displaystyle (\theta^*, v^*) = (\arcsin F, 0), \quad (\pi - \arcsin F, 0)$


**Linearization**

Jacobian

$\qquad\displaystyle A = \begin{pmatrix} 0 & 1 \\ -\cos \theta^* & -2\alpha |v| \end{pmatrix}$

* At $(\theta^*, 0) = (\arcsin F, 0)$:

$\qquad\displaystyle A = \begin{pmatrix} 0 & 1 \\ -\sqrt{1-F^2} & 0 \end{pmatrix}, \quad \lambda_{1,2} = \pm i \sqrt{1-F^2} \quad \text{(center by linearization)}$

* At $(\pi - \arcsin F, 0)$:

$\qquad\displaystyle A = \begin{pmatrix} 0 & 1 \\ \sqrt{1-F^2} & 0 \end{pmatrix}, \quad \lambda_{1,2} = \pm \sqrt{1-F^2} \quad \text{(saddle)}$


**Lyapunov function to confirm nonlinear stability**

Define

$\qquad\displaystyle V(\theta, v) = \frac12 v^2 - F\theta - \cos\theta + F \arcsin F + \sqrt{1-F^2}$

Derivative along trajectories

$\qquad\displaystyle\dot{V} = v \dot{v} - F v + \sin \theta \, v = -\alpha v^2 |v| \le 0$

Conclusion: The fixed point at $(\arcsin F, 0)$ is a **stable spiral**, while $(\pi - \arcsin F, 0)$ is a **saddle**.


## b) Existence and uniqueness of a stable limit cycle

**Construct a trapping region**

$\qquad\displaystyle\dot{v} = F - \alpha v|v| - \sin\theta$

* Upper boundary: $\qquad\displaystyle v = \frac{F+1}{\alpha} \Rightarrow \dot{v} \le 0$
* Lower boundary: $\qquad\displaystyle v = \frac{F-1}{\alpha} \Rightarrow \dot{v} \ge 0$

No fixed points exist inside the strip, so by **Poincaré–Bendixson theorem**, there exists at least one **stable limit cycle**.

**Uniqueness**

Suppose there are two distinct limit cycles $v_U(\theta) > v_L(\theta)$.                                                                            
Integrating the total energy over $0 \le \theta \le 2\pi$ leads to a contradiction

$\qquad\displaystyle \int_0^{2\pi} v_U(\theta) d\theta > \int_0^{2\pi} v_L(\theta) d\theta \quad \Rightarrow \text{contradiction with energy balance.}$

 Therefore, the limit cycle is **unique**.

**Total energy**

$\qquad\displaystyle E = \frac12 v^2 - \cos\theta$

$\qquad\displaystyle\frac{dE}{d\theta} = \frac{dE}{dt} \frac{dt}{d\theta} = \frac{\dot{\theta} \dot{v} + \sin\theta \dot{\theta}}{\dot{\theta}} = \dot{v} + \sin\theta = F - \alpha v|v| $

Integrating over one cycle confirms **existence and uniqueness**.


## c) Change of variables $u = \frac12 v^2$

$\qquad\displaystyle u(\theta) = \frac12 v^2$

$\qquad\displaystyle\frac{du}{d\theta} = v \frac{dv}{d\theta} = v \frac{\dot{v}}{\dot{\theta}} = \dot{v} = \ddot{\theta}$

 This transforms the equation in terms of $\theta$.

## d) Linear equation for $u(\theta)$

For $v > 0$ :

$\qquad\displaystyle\ddot{\theta} + \alpha \dot{\theta}^2 + \sin\theta = F$

$\Rightarrow$ in terms of $u(\theta)$:

$\qquad\displaystyle\frac{du}{d\theta} + 2\alpha u = F - \sin\theta$

This is **linear**, solved with integrating factor :

$\qquad\displaystyle\mu(\theta) = e^{2\alpha \theta}$

$\qquad\displaystyle\frac{d}{d\theta}[u e^{2\alpha \theta}] = (F - \sin\theta) e^{2\alpha \theta}$

**Solution** :

$\qquad\displaystyle u(\theta) = \frac{F}{2\alpha} + \frac{1}{1 + 4 \alpha^2} (\cos\theta - 2\alpha \sin\theta) e^{-2\alpha \theta}$

* The **limit cycle** corresponds to the transient decaying ($e^{-2\alpha \theta} \to 0$):

$\qquad\displaystyle\boxed{u_{\tiny\text{LC}}(\theta) = \frac{F}{2\alpha} + \frac{\cos\theta - 2\alpha \sin\theta}{1 + 4\alpha^2}}$

## e) Homoclinic bifurcation $F_c(\alpha)$

#### Condition for homoclinic bifurcation

At the homoclinic bifurcation:

* The limit cycle **touches the saddle point** $(\theta_s, v_s) = (\pi - \arcsin F, 0)$
* Therefore, at the saddle : $\qquad\displaystyle u(\theta_s) = \frac12 v_s^2 = 0$

* Plug into the limit cycle formula:

$\qquad\displaystyle 0 = \frac{F_c}{2\alpha} + \frac{\cos(\pi - \arcsin F_c) - 2\alpha \sin(\pi - \arcsin F_c)}{1 + 4 \alpha^2}$

$\qquad\displaystyle 0 = \frac{F_c}{2\alpha} + \frac{-\cos(\arcsin F_c) - 2\alpha \sin(\arcsin F_c)}{1 + 4 \alpha^2} \qquad \left[ \ \because \ \cos(\pi - x) = -\cos x, \quad \sin(\pi - x) = \sin x  \right]$

$\qquad\displaystyle 0 = \frac{F_c}{2\alpha} + \frac{-\sqrt{1 - F_c^2} - 2\alpha F_c}{1 + 4 \alpha^2} \qquad \left[ \ \because \ \cos(\arcsin F_c) = \sqrt{1 - F_c^2} , \, \sin(\arcsin F_c) = F_c \right]$

$\qquad\displaystyle 0 = \frac{F_c}{2\alpha} (1 + 4\alpha^2) - \sqrt{1 - F_c^2} - 2 \alpha F_c$

$\qquad\displaystyle 0 = \frac{F_c}{2\alpha} + 2 \alpha F_c - \sqrt{1 - F_c^2} - 2 \alpha F_c = \frac{F_c}{2\alpha} - \sqrt{1 - F_c^2} $

$\qquad\displaystyle \frac{F_c}{2\alpha} = \sqrt{1 - F_c^2} \quad \Rightarrow \quad F_c^2 = 4 \alpha^2 (1 - F_c^2) \quad \Rightarrow F_c^2 + 4\alpha^2 F_c^2 = 4\alpha^2 \quad \Rightarrow \quad F_c^2 (1 + 4\alpha^2) = 4 \alpha^2$

$\qquad\displaystyle \boxed{F_c(\alpha) = \frac{2 \alpha}{\sqrt{1 + 4 \alpha^2}}}$

This is the **exact formula for the homoclinic bifurcation curve**.


**Interpretation**

* For $F > F_c(\alpha)$, there is a **stable limit cycle**.
* At $F = F_c(\alpha)$, the limit cycle collides with the **saddle** → homoclinic bifurcation.
* For $F < F_c(\alpha)$, the limit cycle disappears.


```python
alpha = 0.5
F_c   = 2*alpha / np.sqrt(1 + 4*alpha**2)    # Homoclinic bifurcation
F     = 0.9*F_c                              # Choose slightly below bifurcation for stable limit cycle

theta = np.linspace(-np.pi, 2*np.pi, 500)
v     = np.linspace(-np.pi, np.pi, 500)
Theta, V = np.meshgrid(theta[::25], v[::25])

# Vector field
dTheta = V
dV = F - alpha*V*np.abs(V) - np.sin(Theta)

# Normalize for quiver
N = np.sqrt(dTheta**2 + dV**2) + 1e-6

plt.figure(figsize=(10,6))
plt.quiver(Theta, V, dTheta/N, dV/N, color='blue', alpha=0.5, pivot='mid', scale=20)

# Fixed points
theta_saddle = -np.arcsin(F)
theta_spiral = np.arcsin(F)
plt.plot(theta_saddle, 0, 'rx', markersize=10, label='Saddle')
plt.plot(theta_spiral, 0, 'mo', markersize=10, label='Stable spiral')

u = F/(2*alpha) + (np.cos(theta) - 2*alpha*np.sin(theta))/(1 + 4*alpha**2)
v_lc = np.sqrt(2*u)
plt.plot(theta, v_lc, 'g', lw=2, label='Limit cycle (v>0)')
plt.plot(theta, -v_lc, 'g', lw=2)  # Symmetric v<0 branch

plt.xlim([-np.pi, 2*np.pi])
plt.ylim([-3, 3])
plt.xlabel(r'$\theta$')
plt.ylabel(r'$v = \dot{\theta}$')
plt.title('Quadratically Damped Driven Pendulum Phase Plane\n'+rf'$\alpha={alpha:.2f}, F={F:.2f},F_c={F_c:.2f}$')
plt.legend(loc = 'lower center', fontsize=12)
plt.grid(True)
plt.show()
```

### $Classical \ Dynamics \ of \ Particles \ and \ Systems - S.T.Thornton \ and \ J.B.Marion$

The **driven damped pendulum** is a fundamental example of a nonlinear dynamical system that exhibits a rich variety of behaviors,                         
from simple periodic motion to deterministic chaos. It models a pendulum subject to three forces:

* **Restoring force**: due to gravity, proportional to $\sin x$, where $x$ is the angular displacement,
* **Damping force**: due to friction or air resistance, proportional to the velocity $\frac{dx}{dt}$,
* **Periodic driving force**: an external oscillating force of the form $F \cos(\omega t)$, which supplies energy to the system.

The **equation of motion** for this system is given by:                                                                                                  
$\qquad \displaystyle \frac{d^2 x}{dt^2} + c\frac{d x}{dt} + \sin x = F\cos\left(\omega  t\right) $                                                             
where:

* $x(t)$ is the angular position of the pendulum,
* $c$ is the damping coefficient,
* $F$ is the amplitude of the external driving force,
* $\omega$ is the angular frequency of the driving force.

To analyze the system numerically, we convert this second-order differential equation into a system of **three first-order ODEs** by defining:

* $\displaystyle y = \frac{dx}{dt}$ (angular velocity),
* $\displaystyle z = \omega t$ (phase of the driving force).

The resulting system is:                                                                                                                                   
$ \qquad\displaystyle  \frac{dx}{dt} = y $                                                                                                                                   
$ \qquad\displaystyle  \frac{d y}{dt} = -cy -\sin x  + F \cos z $                                                                                                         
$ \qquad\displaystyle  \frac{dz}{dt} =\omega  $                            


```python
def pendulum_system(t, state, const):
    x, y, z = state   
    c, F, omega = const
    dx_dt = y
    dy_dt = -c * y - np.sin(x) + F * np.cos(z)
    dz_dt = omega
    return [dx_dt, dy_dt, dz_dt]
```

### **Parametric pendulum exhibiting tumbling chaotic motion**

For a pendulum of length $L$ and angle $x$ (or $\theta$) :

$\qquad \displaystyle\ddot{x}+\frac{g}{L}\sin x=0$

define the natural frequency $\omega_0^2=\frac{g}{L}$

So the equation becomes

$\qquad \displaystyle\ddot{x}+\omega_0^2\sin x=0$

Real pendulums experience friction proportional to velocity $(h\dot{x})$

Thus the **damped pendulum** becomes

$\qquad \displaystyle \ddot{x}+h\dot{x}+\omega_0^2\sin x=0 $

Suppose the **effective gravitational term is periodically modulated**                                                                                       
(for example, the pivot oscillates vertically or the length varies).

Then

$\qquad \displaystyle\omega_0^2 \rightarrow \omega_0^2+\varepsilon\cos(\omega t)$

Thus the restoring torque becomes time dependent.

The equation becomes

### $\qquad \displaystyle \boxed{\ddot{x}+h\dot{x}+\left[\omega_0^2+\varepsilon\cos(\omega_{\tiny D} t)\right]\sin x=0 \ }$



```python
def pendulum_system(t, state, const):
    tau, x, y = state
    h, q, w_D, w0 = const
    dtau = 1
    dx   = y
    dy   = - h*y - (w0**2 + q*np.cos(w_D*tau)) * np.sin(x)
    return [dtau, dx, dy]
```

```python
h = 0.1
q = 2.0
w = 2.0
w0 = 1.0
T  = 2*np.pi / w
x0 = [0.0, 0.1, 0.0]
dt = 0.01
const = (h, q, w, w0)
t_span = (0, 50000)

t, sol = RK4(pendulum_system, t_span, const, dt, x0)
t_vals,x_vals,y_vals = sol[:,0],sol[:,1],sol[:,2]
strobe_times = np.arange(0, t_span[1], T)
indices = [np.argmin(np.abs(t - s)) for s in strobe_times]
x_poincare = np.mod(x_vals[indices] + np.pi, 2*np.pi) - np.pi
y_poincare = y_vals[indices]

fig, ax = plt.subplots( figsize=(6,4))
ax.plot(x_poincare[100:], y_poincare[100:], 'r,', markersize=2)
ax.set_title("Poincaré Section")
ax.set_xlim(x_poincare.min(), x_poincare.max())
ax.set_xticks([-np.pi, 0, np.pi], [r'$-\pi$',r'$0$',r'$\pi$'])
ax.set_xlabel(r'$ x \ \mathrm{ mod } \ (2\pi k) $')
ax.set_ylabel(r'$ y \ \mathrm{ mod } \ (2\pi k)$')
plt.tight_layout()
plt.show()
```
